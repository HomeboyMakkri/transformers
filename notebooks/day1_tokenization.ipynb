## D1-01 — Language, checkpoint, and format

We use English text and `distilbert-base-uncased`. The tokenizer lowercases input because this is an uncased checkpoint.

```python
print(f"Model name: {DEFAULT_MODEL_NAME}")
tokenizer = load_tokenizer()
```

## D1-02 — Inspect the tokenizer

The vocabulary maps known tokens to IDs. `model_max_length` is the maximum supported sequence length, including special tokens. `[CLS]` starts the sequence, `[SEP]` closes it, `[PAD]` aligns shorter examples in a batch, and `[UNK]` represents text the vocabulary cannot decompose.

```python
info = inspect_tokenizer(tokenizer)
print(f"Vocabulary size: {info.vocabulary_size}")
print(f"Maximum length: {info.model_max_length}")
for name, token in info.special_tokens.items():
    print(f"{name}: {token!r}, id={info.special_token_ids[name]}")
```

## D1-03 — One text from characters to IDs and back

WordPiece may split a word into several vocabulary pieces. Encoding adds special tokens. Decoding reconstructs a normalized representation, so casing, whitespace, and punctuation spacing need not exactly match the original input.

```python
text = "This movie was absolutely amazing!"
walkthrough = tokenize_single_text(text, tokenizer)
print(f"Original: {walkthrough.text}")
print(f"Tokens ({len(walkthrough.tokens)}): {walkthrough.tokens}")
print(f"Input IDs: {walkthrough.input_ids}")
print(f"Attention mask: {walkthrough.attention_mask}")
print(f"Decoded: {walkthrough.decoded_text}")
```
